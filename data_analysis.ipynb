Короткий опис ознак набору даних (надано вручну)
id — унікальний ідентифікатор студента (штучний індекс, не несе аналітичного навантаження)
Gender — стать респондента (Категоріальна ознака: Male / Female)
Age — вік студента у роках (Числова ознака)
Citу — місто проживання або навчання (Категоріальна ознака)
Profession — рід занять (у межах цього датасету всі записи відповідають значенню Student)
Work Pressure — рівень робочого тиску (Числова/Рангова ознака; для студентів, які не працюють, дорівнює 0)
CGPA — середній бал успішності студента (Числова безперервна ознака)
Study Satisfaction — рівень задоволеності процесом навчання (Числова/Рангова ознака від 0 до 5)
Job Satisfaction — рівень задоволеності роботою (Числова/Рангова ознака; для безробітних студентів дорівнює 0)
Sleep Duration — тривалість сну на добу (Категоріальна/Інтервальна ознака)
Dietary Habits — характер та якість харчування (Категоріальна ознака: Healthy / Moderate / Unhealthy)
Degree — освітній ступінь або спеціалізація, за якою навчається студент (Категоріальна ознака)
Have you ever had suicidal thoughts ? — наявність суїцидальних думок в анамнезі (Бінарна ознака: Yes / No)
Work/Study Hours — кількість годин, що витрачаються на навчання чи роботу щодня (Числова ознака)
Financial Stress — рівень стресу, пов'язаний із фінансовим становищем (Числова/Рангова ознака від 0 до 5)
Family History of Mental Illness — наявність психічних захворювань у сімейному анамнезі (Бінарна ознака: Yes / No)
Depression — наявність діагностованої депресії (Цільова бінарна ознака: 1 — так, 0 — ні)

таблиця з абсолютними та відносними значеннями

```python
for col in cat_cols:
    print(f"Аналіз розподілу для ознаки: {col} ")
    
    counts = data[col].value_counts()
    percentages = data[col].value_counts(normalize=True) * 100
    
    distribution_df = pd.DataFrame({
        'Кількість (абс.)': counts,
        'Частка (%)': percentages.round(2)
    })
    
    print(distribution_df)
    print("" + "="*50 + "")
```

```python
print(" Розподіл цільової змінної 'Depression'")
print(data['Depression'].value_counts(normalize=True) * 100) 
```

переіменування колонок

```python
data = data.rename(columns={
    'Have you ever had suicidal thoughts ?': 'Suicidal_Thoughts',
    'Family History of Mental Illness': 'Family_History',
    'Academic Pressure': 'Academic_Pressure',
    'Work Pressure': 'Work_Pressure',
    'Study Satisfaction': 'Study_Satisfaction',
    'Job Satisfaction': 'Job_Satisfaction',
    'Sleep Duration': 'Sleep_Duration',
    'Dietary Habits': 'Dietary_Habits',
    'Work/Study Hours': 'Work_Study_Hours',
    'Financial Stress': 'Financial_Stress'
})
```

```python
data.sample(10)
```

видалення непотрібної колонки

```python
data = data.drop(columns=['id', 'Job_Satisfaction', 'Work_Pressure'])
```

переводимо yes/no у 1/0

```python
binary_map = {'Yes': 1, 'No': 0}
data['Suicidal_Thoughts'] = data['Suicidal_Thoughts'].map(binary_map)
data['Family_History'] = data['Family_History'].map(binary_map)
```

загальний рівень стресу

```python
data['Total_Pressure'] = (
    data['Academic_Pressure'].fillna(0) +  
    data['Financial_Stress'].fillna(0)
)
```

групуємо вік 

```python
bins = [0, 20, 25, 30, 40, 100]
labels = ['<20', '21-25', '26-30', '31-40', '40+']
data['Age_Group'] = pd.cut(data['Age'], bins=bins, labels=labels)
```

особливості датасету: розмір великий (27 901 рядків, спочатку було 18 колонок). структура поєднує числові й категоріальні дані про студентів. незвичний момент — стовпеці Job Satisfaction та Work Pressure повністю заповнені нулями 0.0, оскільки всі опитані мають професію Student.

поля, що потребували найбільшої уваги: Financial Stress (єдиний стовпець із пропусками — 3 значення) та текстові ознаки з відповідями Yes/No (Suicidal_Thoughts, Family_History), які треба було закодувати в 1/0 для нормальної роботи алгоритмів. Переважна більшість математичних алгоритмів класифікації (лінійні моделі, метричні методи, нейронні мережі) оперують виключно матрицями чисел і не здатні безпосередньо приймати текстові рядки як вхідні параметри (features).

колонки без корисної інформації: id (просто штучний індекс), Work Pressure та Job Satisfaction (має нульову дисперсію, бо там лише нулі) успішно видалено з датасету в коді.

обробка пропущених значень та обґрунтування вибору медіани:
для відновлення пропущених значних у числових стовпцях було застосовано заповнення медіаною*, а не середнім арифметичним. 
математичне обґрунтування: медіана є стійкою метрикою центральної тенденції. На відміну від середнього арифметичного, яке суттєво зміщується під впливом асиметрії розподілу або поодиноких статистичних викидів (екстремально великих чи малих значень), медіана відображає реальний "центр" даних. Це дозволяє мінімізувати штучне спотворення розподілу ознак.

закономірності на цьому етапі: цільова змінна Depression відносно збалансована (58.5% — так, 41.5% — ні). основна маса студентів хронічно недосипає (найчастіші значення сну — менше 5 годин або 5-6 годин), що явно корелює із загальним рівнем депресії.

труднощі при обробці: серйозних проблем не було, датасет чистий. робота звелася до виправлення синтаксису комірок, перейменування довгих назв, мапінгу бінарних значень та групування віку.

готовність до аналізу та моделювання: дані повністю готові. пропуски закриті медіаною, зайві колонки прибрано, категорії переведено в числа. створені похідні ознаки (Total_Pressure та Age_Group) готові для побудови графіків.
ЗАГАЛЬНИЙ АНАЛІЗ
gender (стать): розподіл відносно збалансований (55.7% — чоловіки, 44.3% — жінки).

city (місто): у хвостах розподілу є поодинокі помилки введення та сміттєві значення (імена, випадкові числа, назви ступенів), їх менше 0.1%.

profession (професія): 99.89% записів мають значення student. ознака фактично константна і не несе корисної інформації для аналізу, тому її разом із work pressure та job satisfaction можна видалити.

sleep duration (сон): явна проблема з недосипанням, понад 50% студентів сплять 6 годин або менше (при цьому 29.8% — менше 5 годин).

dietary habits (харчування): лише 27.4% харчуються здорово, більшість віддає перевагу нездоровому (37%) або помірному (35.6%) харчуванню.

degree (ступінь): найбільша група це class 12 (21.8%), що вказує на велику кількість абітурієнтів або першокурсників у вибірці.

suicidal thoughts (суїцидальні думки): високий показник (63.3% відповідей так), ця колонка є потенційно найсильнішим предиктором для моделювання депресії.

family history (сімейний анамнез): вибірка розділилася майже навпіл (51.6% — ні, 48.4% — так), що ідеально підходить для аналізу спадкового фактору.
